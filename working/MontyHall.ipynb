# Monty Hall

```python
import numpy as np
```

```python
N = [10,1000,1000000]

for n in N:
    
    v_switcher = 0
    v_conservative = 0
    v_newcomer = 0
    print('Caso ',n,'\n')

    for i in range(n):
        
        porte = np.array([0,0,1])
        scelta = np.random.randint(0, 3)
        #print ('Porta scelta: ',scelta)
    
        possibili = np.where((porte == 0) & (np.arange(3) != scelta))[0]
        elimina = np.random.choice(possibili)
        
        new = [j for j in range(3) if j not in [scelta, elimina]][0]
    
        if porte[new] == 1:
            v_switcher += 1
    
        scelta_new = np.random.choice([scelta, new])
        if porte[scelta_new] == 1:
            v_newcomer += 1
    
        if porte[scelta] == 1:
            v_conservative +=1
             
    print ('Vittorie switcher: ',v_switcher)
    print ('Vittorie newcomer: ',v_newcomer)
    print ('Vittorie conservative: ',v_conservative)
    
    print('Percentuali:\n - Switcher: ',v_switcher/n,'%\n - Conservative ',v_conservative/n,'%\n - New comer ',v_newcomer/n,'%')

    
```

# N porte

```python

N = 10000
v_switcher = 0
v_conservative = 0
v_newcomer = 0
n_porte = 10
aperte = 8
porte_aperte = []

for i in range(N):
    porte_aperte = []
    porte = np.concatenate((np.zeros(n_porte-1), [1]))
    np.random.shuffle(porte)  
    scelta = np.random.randint(0, len(porte))
   
    for x in range(aperte):
        possibili = np.where((porte == 0) & (np.arange(len(porte)) != scelta))[0]
        elimina = np.random.choice(possibili)
        porte_aperte.append(elimina)

    #print(porte_aperte)
    arr = np.array(porte_aperte)
   

    new = [j for j in range(len(porte)) if j not in [scelta, arr.all()]][0]
    #print(new)
    
    if porte[new] == 1:
        v_switcher += 1

    scelta_new = np.random.choice([scelta, new])
    if porte[scelta_new] == 1:
        v_newcomer += 1

    if porte[scelta] == 1:
        v_conservative +=1
         
print ('Vittorie switcher: ',v_switcher)
print ('Vittorie newcomer: ',v_newcomer)
print ('Vittorie conservative: ',v_conservative)

print('Percentuali:\n - Switcher: ',v_switcher/N,'%\n - Conservative ',v_conservative/N,'%\n - New comer ',v_newcomer/N,'%')

    
```
