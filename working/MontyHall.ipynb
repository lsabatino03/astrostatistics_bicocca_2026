# Monty Hall

```python
import numpy as np
```

```python
porte = np.array([0,0,1])
print(porte)
N = 10000
v_switcher = 0
v_conservative = 0

#Switcher
for i in range(N):
    porte = np.array([0,0,1])
    scelta = np.random.randint(0, 3)
    #print ('Porta scelta: ',scelta)

    if porte[scelta] == 1:
        v_switcher +=1
    else:
        elimina = np.random.randint(0,2)
        #print('Porta eliminata: ',elimina)
        porte = np.delete(porte, elimina)

        new = np.random.randint(0,2)
        if porte[new]==1:
            v_switcher +=1
print ('Vittorie switcher: ',v_switcher)

#Conservative
for i in range(N):
    porte = np.array([0,0,1])
    scelta = np.random.randint(0, 3)
    #print ('Porta scelta: ',scelta)

    if porte[scelta] == 1:
        v_conservative +=1
    else:
        elimina = np.random.randint(0,2)
        #print('Porta eliminata: ',elimina)
        porte = np.delete(porte, elimina)

        new = scelta
        if porte[new]==1:
            v_conservative +=1
print ('Vittorie conservative: ',v_conservative)
```
