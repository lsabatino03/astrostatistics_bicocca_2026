# Pdf of the log

```python
import numpy as np
from matplotlib import pyplot as plt
```

```python
import numpy as np
import matplotlib.pyplot as plt

N = 10000
x = np.random.uniform(0.1, 10, size=N)
valori = np.linspace(np.log10(0.1), np.log10(10), 200)
py = np.log(10)*10**valori/9.9


plt.figure()
plt.hist(x, density = True)
plt.xlabel('x')
plt.ylabel('Frequenza')
plt.title('Istogramma di x')


y = np.log10(x)
plt.figure()
plt.hist(y, density = True)
plt.xlabel('log10(x)')
plt.ylabel('Frequenza')
plt.title('Istogramma di log10(x)')
plt.plot(valori, py, label =r'$pdf_y = log_{10}10^{y}/9.9$')
plt.legend()

mediax = np.mean(x)
mediax_log = np.log10(mediax)
print(f'Log mean x: {mediax_log:.4f}')

mediay = np.mean(y)
print(f'Log mean y: {mediay:.4f}')

medianx = np.median(x)
medianx_log = np.log10(medianx)
print(f'\nLog median x: {medianx_log:.4f}')

mediany = np.median(y)
print(f'Log median y: {mediany:.4f}')




plt.show()
```
