```python
import numpy as np
import matplotlib.pyplot as plt

# Parametri
N = 100000      # numero di campioni
a = 0.1         # limite inferiore
b = 10          # limite superiore

# Genero i campioni uniformi
x = np.random.uniform(a, b, N)

# Istogramma di x
plt.figure(figsize=(12,5))

plt.subplot(1,2,1)
plt.hist(x, bins=50, density=True, color='skyblue', edgecolor='black')
plt.xlabel('x')
plt.ylabel('Densità')
plt.title('Istogramma di x (Uniforme)')
plt.grid(True)

# Log10 dei campioni
y = np.log10(x)

# Istogramma di y
plt.subplot(1,2,2)
plt.hist(y, bins=50, density=True, color='salmon', edgecolor='black', alpha=0.6, label='Istogramma simulato')

# PDF teorica di y
y_vals = np.linspace(np.log10(a), np.log10(b), 200)
pdf_y = (np.log(10) * 10**y_vals) / (b - a)

plt.plot(y_vals, pdf_y, 'k-', lw=2, label='PDF teorica')
plt.xlabel('y = log10(x)')
plt.ylabel('Densità')
plt.title('Istogramma di log10(x) con PDF teorica')
plt.legend()
plt.grid(True)

plt.tight_layout()
plt.show()
```
